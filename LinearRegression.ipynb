```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5]]
y = [3, 5, 7, 9, 11]

model = LinearRegression()

model.fit(X, y)

print(model.coef_[0])
print( model.intercept_)

print("Prediction:", model.predict([[6]]))
```

Multiple Features

```python
from sklearn.linear_model import LinearRegression

x = [
    [1, 1],
    [2, 1],
    [1, 2],
    [2, 2]
]

y = [4, 6, 5, 7]

model = LinearRegression()

model.fit(x, y)

print("w1 =", model.coef_[0])
print("w2 =", model.coef_[1])
print("w0 =", model.intercept_)

prediction = model.predict([[3, 4]])
print("Prediction =", prediction[0])
model.score(x,y)
```
