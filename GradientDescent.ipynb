```python
#prediction = w0 + w1x
#Uni Variate

x = [1,2,4,3,5]
y = [1,3,3,2,5]

learning_rate = 0.01
w0 = 0
w1 = 0
itr = 10

for i in range(itr):
    predictions = []
    for val in x:
        p = w0 + w1 * val
        predictions.append(p)
    gra0 = 0
    gra1 = 0

    for j in range(len(x)):
        gra0+= predictions[j] - y[j]
        gra1+= (predictions[j] - y[j]) * x[j]
    gra0 = gra0 / len(x)
    gra1 = gra1 / len(x)

    w0 = w0 - learning_rate * gra0
    w1 = w1 - learning_rate * gra1


    print("Iteration: ", i+1, "w0: ", round(w0,4), "w1: ", round(w1,4))
            
            
```

```python
#prediction = w0 + w1x + w2x2
#Multi Variate

x = [ [1, 1], [2, 1], [1, 2], [2, 2] ]
y = [3, 5, 4, 6]

learning_rate = 0.01
w0 = 0
w1 = 0
w2 = 0
itr = 10

for i in range(itr):
    predictions = []
    for val in x:
        p = w0 + w1 * val[0] + w2 * val[1]
        predictions.append(p)
        
    gra0 = 0
    gra1 = 0
    gra2 = 0

    for j in range(len(x)):
        gra0+= predictions[j] - y[j]
        gra1+= (predictions[j] - y[j]) * x[j][0]
        gra2+= (predictions[j] - y[j]) * x[j][1]
    gra0 = gra0 / len(x)
    gra1 = gra1 / len(x)
    gra2 = gra2 / len(x)

    w0 = w0 - learning_rate * gra0
    w1 = w1 - learning_rate * gra1
    w2 = w2 - learning_rate * gra2


    print("Iteration: ", i+1, "w0: ", round(w0,4), "w1: ", round(w1,4), "w2: ", round(w2,4))
```

```python
from sklearn.linear_model import SGDRegressor

x = [[1], [2], [3], [4], [5]]
y = [1, 3, 2, 3, 5]


model = SGDRegressor(
    learning_rate="constant",
    eta0=0.01,
    max_iter=1000,
    random_state=56
)
model.fit(x, y)

# w1
print("w1:", model.coef_)

# w0
print("w0:", model.intercept_[0])

# Prediction for x = 5
prediction = model.predict([[5]])

print("Prediction:", prediction[0])
```

```python
from sklearn.linear_model import SGDRegressor

# Dataset
X = [[1, 1],[2, 1],[1, 2],[2, 2]]

y = [3, 5, 4, 6]


model = SGDRegressor(
    learning_rate="constant",
    eta0=0.01,
    max_iter=1000,
    tol=None,
    random_state=42
)


model.fit(X, y)


print("w1:", model.coef_[0])
print("w2:", model.coef_[1])

# Bias
print("w0:", model.intercept_)

# Prediction
prediction = model.predict([[3, 2]])

print("Prediction:", prediction[0])
```
