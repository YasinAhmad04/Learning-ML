```python
from sklearn.cluster import KMeans
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from matplotlib import pyplot as plt
%matplotlib inline

dataset = pd.read_csv(r'D:\Abdullah\IRIS.csv')
df = dataset.drop(['sepal_length','sepal_width','species'], axis='columns')

km = KMeans(n_clusters=3)
ans = km.fit_predict(df)
# ans

df['cluster'] = ans
df.cluster.unique()
#print(df)
#df

df1 = df[df.cluster==0]
df2 = df[df.cluster==1]
df3 = df[df.cluster==2]



plt.scatter(df1['petal_length'],df1['petal_width'],color='blue')
plt.scatter(df2['petal_length'],df2['petal_width'],color='green')
plt.scatter(df3['petal_length'],df3['petal_width'],color='yellow')

```
